```python
from sqlalchemy import create_engine, text, inspect
```

```python
engine = create_engine('sqlite:///data/scifi_books_hugo_awarded.db')
```

```python
with engine.begin() as conn:
    table = conn.execute(text("SELECT * FROM books_awarded LIMIT 5"))
    print(table.all()[0])
```

```python
import csv
writer = csv.writer(open("data/scifi_books_hugo_awarded.csv", 'w'))
with engine.begin() as conn:
    table = conn.execute(text("SELECT * FROM books_awarded LIMIT 5"))
    for row in table:
        writer.writerow([row[0], row[1], row[2], row[3], row[4]])
```

```python
csv_file = "data/scifi_books_hugo_awarded.csv"
reader = csv.reader(open(csv_file, 'r'))
for row in reader:
    print(row)
```
