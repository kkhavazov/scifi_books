```python
import sqlite3
conn = sqlite3.connect("data/scifi_books.db")
cursor = conn.cursor()
cursor.execute("ATTACH DATABASE 'data/scifi_books_hugo_awarded.db' AS db2")
query = """
    SELECT t1.id, t2.title, t2.winner
    FROM main.books AS t1
    RIGHT JOIN db2.books_awarded AS t2 ON LOWER(t1.Book_Title) = LOWER(t2.title)
    WHERE t2.category = 'Novel' 
"""

cursor.execute(query)
results = cursor.fetchall()
for row in results:
    print(row)
conn.close()
```

```python
import sqlite3

conn = sqlite3.connect("data/scifi_books.db")
cursor = conn.cursor()

cursor.execute("ATTACH DATABASE 'data/scifi_books_hugo_awarded.db' AS award_db")

cursor.execute(
    "CREATE TABLE main.books_awarded AS SELECT * FROM award_db.books_awarded"
)

conn.commit()
conn.close()

print("Table successfully copied!")
```
