```python
import pandas as pd
from os import listdir
from os.path import isfile, join
import ast
```

```python
mypath = "data/books_genres"
onlyfiles = [f for f in listdir(mypath) if isfile(join(mypath, f)) and f.endswith(".csv")]
```

```python
onlyfiles
```

```python
columns = ['Book_Title', 'Original_Book_Title', 'Author_Name', 'Edition_Language',
       'Rating_score', 'Rating_votes', 'Review_number', 'Book_Description',
       'Year_published', 'Genres', 'url', 'Subgenre']
```

```python
df = pd.DataFrame(columns=columns)
for file in onlyfiles:
    print(file)
    df_subgenre = pd.read_csv(join(mypath, file))
    subgenre = file[3:-4]
    df_subgenre["Subgenre"] = subgenre
    df = pd.concat([df, df_subgenre], ignore_index=True)
print(df.head())
```

```python
df = df.drop_duplicates(subset=["Book_Title", "Author_Name"], keep="first")
df.sort_values(by="Review_number", ascending=False)
```

```python
df.to_csv("data/scifi_books.csv", index=False)
```

```python
for row in df.iterrows():
    genres = ast.literal_eval(row[1]["Genres"])
    summary = sum(genres.values())
    new_row = {}
    for genre in genres:
        ratio = genres[genre] / summary
        new_row[genre] = ratio
    row[1]["Genres"] = new_row
    print(row[1]["Genres"])
```

```python
def calculate_genre_ratios(genres_str):
    genres = ast.literal_eval(genres_str)
    
    total = sum(genres.values())
    
    if total == 0:
        return {genre: 0.0 for genre in genres}
    return {genre: f"{count / total:.2f}" for genre, count in genres.items()}
df['Genres'] = df['Genres'].apply(calculate_genre_ratios)
```
